```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import PeftModel
```

```python
BASE_MODEL = "google/gemma-2b-it"
ADAPTER_MODEL = "" # 학습 후 모델저장 경로
```

```python
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)
```

## 학습 전 모델 테스트

```python
model_origin = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)
```

```python

# 추론 모드
model_origin.eval()

tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)
tokenizer.pad_token = tokenizer.eos_token

# 테스트 문서
document = """
서울시는 29일 2025년 예산안을 발표했다. 총 예산 규모는 50조원으로 
전년 대비 5% 증가했다. 주요 투자 분야는 교통 인프라 확충과 
복지 정책 강화다. 시는 지하철 노선 확장과 저소득층 지원 예산을 
대폭 늘렸다고 밝혔다.
"""

# 프롬프트 생성
prompt = f"""<bos><start_of_turn>user
다음 글을 요약해주세요:

{document.strip()}<end_of_turn>
<start_of_turn>model
"""

# 토크나이징
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# 생성
print("\n생성 중...")
with torch.no_grad():
    outputs = model_origin.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.7,
        top_p=0.9,
        do_sample=True,
        pad_token_id=tokenizer.pad_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )

# 결과 출력
result = tokenizer.decode(outputs[0], skip_special_tokens=False)
print("\n=== 전체 출력 ===")
print(result)

# 요약만 추출
summary = result.split("<start_of_turn>model")[-1].replace("<end_of_turn>", "").replace("<eos>", "").strip()
print("\n=== 요약 결과 ===")
print(summary)
```
